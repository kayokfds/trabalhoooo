##Leitura de Dados

```python
nomes_arquivos = [
    'COTAHIST_A2025.TXT',
    'COTAHIST_A2024.TXT',
    'COTAHIST_A2023.TXT',
    'COTAHIST_A2022.TXT',
    'COTAHIST_A2021.TXT',
    'COTAHIST_A2020.TXT',
    'COTAHIST_A2019.TXT',
    'COTAHIST_A2018.TXT',
    'COTAHIST_A2017.TXT',
    'COTAHIST_A2016.TXT'
]
```

```python
colspecs = [
    (0, 2), (2, 10), (10, 12), (12, 24), (24, 27), (27, 39), (39, 49),
    (49, 52), (52, 56), (56, 69), (69, 82), (82, 95), (95, 108), (108, 121),
    (121, 134), (134, 147), (147, 152), (152, 170), (170, 188), (188, 201),
    (201, 202), (202, 210), (210, 217), (217, 230), (230, 242), (242, 245),
]

col_origem = [
    "tipreg", "data", "codbdi", "codneg", "tpmerc", "nomres", "especi",
    "prazot", "modref", "preabe", "premax", "premin", "premed", "preult",
    "preofc", "preofv", "totneg", "quatot", "voltot", "preexe", "indopc",
    "datven", "fatcot", "ptoexe", "codisi", "dismes",
]

colunas = [
    "data", "codbdi", "codneg", "tpmerc", "nomres", "especi",
    "preabe", "premax", "premin", "premed", "preult",
    "totneg", "quatot", "voltot", "fatcot",
]

dados = {}
for i in range(len(nomes_arquivos)):
  arquivo = nomes_arquivos[i]
  aaa = pd.read_fwf(
            f"{caminho_dados}/{arquivo}",
            colspecs=colspecs,
            names=col_origem,
            dtype=str,
            encoding="latin-1"
        )
  aaa = aaa[colunas]
  dados[i] = aaa
  print(f"Arquivo lido: {arquivo}; Tamanho: {aaa.shape}")
print("Funcionou!!")
```

```python
df_analisado = 0

dados[0].head()
```

```python
base_completa = pd.DataFrame(columns = ["origem"]+colunas)
base_completa
```

```python
lista_dfs = []

for i in dados:
    df = dados[i]
    df["origem"] = df.iloc[0, 3]
    df = df[df['data'] != 'COTAHIST']
    lista_dfs.append(df)
    # dados[i] = "Concluído"

base_completa = pd.concat(lista_dfs, ignore_index=True)
del lista_dfs, df
```

```python
print(base_completa.shape, base_completa.dtypes)
base_completa.head()
```

```python
## Data
base_completa['data'] = pd.to_datetime(base_completa['data'], format='%Y%m%d')

## Índices e Valores
numericos = ['preabe', 'premax', 'premin', 'premed', 'preult', 'totneg', 'quatot', 'voltot', 'fatcot']
base_completa[numericos] = base_completa[numericos].apply(pd.to_numeric, errors='coerce')
base_completa[['preabe', 'premax', 'premin', 'premed', 'preult']] = base_completa[['preabe', 'premax', 'premin', 'premed', 'preult']] / 100
```
